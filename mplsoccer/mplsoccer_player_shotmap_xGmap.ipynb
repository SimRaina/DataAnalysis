# 📊 Football Visualizations using `mplsoccer`
This notebook demonstrates how to create football data visualizations — including shot maps, xG maps — using Erling Haaland's 2022 Understat data.

```python
# Import libraries
import pandas as pd
from mplsoccer import Pitch
import matplotlib.pyplot as plt
```

```python
# Read CSV file
df = pd.read_csv('erling_haaland_2022_understat.csv')
df.head()
```

```python
# Prepare coordinates (convert Understat to StatsBomb scale)
df['x'] = df['X'] * 120
df['y'] = 80 - (df['Y'] * 80)
```

## ⚽ Shot Map

```python
pitch = Pitch(pitch_type='statsbomb', pitch_color='grass', line_color='white')
fig, ax = pitch.draw(figsize=(8, 6))

goals = df[df['result'] == 'Goal']
misses = df[df['result'] != 'Goal']

pitch.scatter(misses['x'], misses['y'], ax=ax, s=60, edgecolors='black', c='red', label='Missed')
pitch.scatter(goals['x'], goals['y'], ax=ax, s=80, edgecolors='black', c='gold', label='Goal')

ax.legend()
ax.set_title('Erling Haaland – Shot Map (2022)', fontsize=14)
plt.show()
```

## 📈 xG Map

```python
fig, ax = pitch.draw(figsize=(8, 6))
sc = pitch.scatter(df['x'], df['y'], s=df['xG']*800, c=df['xG'], cmap='coolwarm', alpha=0.8, ax=ax)
plt.colorbar(sc, ax=ax, label='xG')
ax.set_title('Haaland xG Map (2022)', fontsize=14)
plt.show()
```
